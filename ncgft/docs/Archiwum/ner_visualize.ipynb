```python
import spacy
from spacy import displacy
from spacy.tokens import Doc


nlp = spacy.blank('en')

doc = Doc(nlp.vocab,
        words=['Right', '7', 'mm', 'nodule', 'in', 'the', 'right', 'upper', 'lobe', 'noted', 'on', 'CT','.'],
        spaces=[True, True, True, True, True, True, True, True, True, True, True, True, False])

annotations_char_spans = [(0, 5, 'laterality'), (6, 10, 'size'), (11, 17, 'lesion'),(25,41,'location')]

custom_entitites = []
for span_start, span_end, label in annotations_char_spans:
    ent = doc.char_span(span_start, span_end, label=label)
    if ent is None:
        continue
    custom_entitites.append(ent)
doc.ents = custom_entitites

options = {"ents": ['lesion', 'size', 'location', 'laterality'],
           "colors": {'lesion': "#85C1E9", "size": "#FF6961", "location": "#77DD77","laterality":"#F49AC2"}}
displacy.serve(doc, style="ent", host="localhost", port=6060, options=options)
```
